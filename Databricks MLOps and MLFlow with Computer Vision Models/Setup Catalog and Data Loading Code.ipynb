# Data Loading

- The data can be loaded through UI of databricks. But, after facing continuous error can also be resorted through code using databricks.sdk.

```python
# Run on local computer (NOT DATABRICKS) after downloading dataset

import os
from pathlib import Path
from databricks.sdk import WorkspaceClient


# 1. Insert databricks url
DATABRICKS_HOST = "demo"

# 2. Insert databricks personal access token
DATABRICKS_TOKEN = "demo"

# 3. Path to local folder on the computer
LOCAL_FOLDER = "./Users/folder" 

# 4. Destination path in Databricks
REMOTE_FOLDER = "/Volumes/catalog/schema/"


def upload_directory_to_databricks():
    # Initialize with credentials
    w = WorkspaceClient(
        host=DATABRICKS_HOST,
        token=DATABRICKS_TOKEN
    )
    
    local_path = Path(LOCAL_FOLDER)

    total_files = 0
    uploaded_files = 0

    for root, _, files in os.walk(local_path):
        for file_name in files:
            total_files += 1
            file_path = Path(root) / file_name
            
            relative_path = file_path.relative_to(local_path).as_posix()
            dest_file_path = f"{REMOTE_FOLDER.rstrip('/')}/{relative_path}"
            
            with open(file_path, "rb") as f:
                w.files.upload(
                    file_path=dest_file_path,
                    contents=f,
                    overwrite=True
                )
            uploaded_files += 1

if __name__ == "__main__":
    upload_directory_to_databricks()
```
